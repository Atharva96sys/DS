##### ***Ananlyzing Data from data frames***

Q1.What is the total number of reported cases and deaths related to covid-19 in italy?

```python
total_cases = covid_df['new_cases'].sum()
print(total_cases)
```

```python
total_deaths = covid_df['new_deaths'].sum()
print(total_deaths)
```

```python
death_rate = (total_deaths/total_cases)*100

print("The overall death rate in Italy is {:.2f} %".format(death_rate))
```

Q2.What is the overall number of test conducted?.A total of 935310 test were conducted before daily test numbers were beign reported.

```python
initial_tests = 935310
total_tests = initial_tests + covid_df['new_tests'].sum()
print("Total Number of tests conducted in italy are {:.2f} ".format(total_tests))
```

Q3.What fraction of tests returned a positive result?

```python
tests_positive_rate = total_cases/total_tests
tests_positive_rate
print("Percentage of successful positive test result is {:.2f}%".format(tests_positive_rate*100))
```
